# Associative networks

In this lab, we will implement the associative network introduced in class. As discussed in lecture, we consider a network architecture where a set of input nodes projects directly to a set of output nodes, without any hidden layers:

trinket_plot.png

This schematic network consists of 4 input nodes and 4 output nodes, where each node can take on different values (represented by different shades of red in our visualization). Mathematically, we can think of the input here as a vector $\vec{x} \in \mathbb{R}^4$ and the output as a vector $\vec{y} \in \mathbb{R}^4$. Each component of these vectors represents the activation value of the corresponding node.

The key question we're thinking about is: how can we transform an input vector $\vec{x}$ into the desired output vector $\vec{y}$? The transformation is accomplished through a weight matrix $\textbf{A}$ that represents the strength of connections between input and output nodes. Each line in our diagram corresponds to a weight $a_{ij}$ in this matrix, representing how strongly input node $j$ influences output node $i$.

# Task 1 - A first association

The first task for you is to develop a simple code that creates a network that is able to store a simple association, and then check if the network is sucessufuly recovering the associated memory.

# Solution

```python
import numpy as np

# parameters
N = 5
eta = 1

# input vector (x)
x = 2 * (np.random.rand( N, 1) - 0.5 );
x = x / np.linalg.norm(x) # unit norm

# output vector (y)
y = 2 * (np.random.rand( N, 1) - 0.5 );
y = y / np.linalg.norm(y) # unit norm

# Hebbian outer product
A = eta * np.outer( y, x )

# compute output
y_prime = A @ x

# output inner product of y and y'
print( y.T @ y_prime)
```

# Task 2 - Geometry of random vectors in high dimensions

The second task for you is to develop a simple code to test the orthogonality of random vectors in high dimensions. One of the assumptions we consider in our lecture was the fact that random vectors can be quasi-orthogonals in high dimensions. Let's now test this numerically. In our code, vary the dimension and check how it affects the orthogonality of your random vectors.

# Solution

```python
import numpy as np
import matplotlib.pyplot as plt

# parameters
N = np.arange(2, 10000, step=1 )
d = np.zeros( len(N) )

for ii in np.arange(0, len(N)):
  x1 = 2 * ( np.random.rand( N[ii], 1 ) - 0.5)
  x1 = x1 / np.linalg.norm( x1 )
  x2 = 2 * ( np.random.rand( N[ii], 1 ) - 0.5)
  x2 = x2 / np.linalg.norm( x2 )
  d[ii] = ( np.matmul( np.transpose(x1), x2 ) )

plt.plot( N, d, 'k.' )
plt.xlabel( 'N' )
plt.show()
```
